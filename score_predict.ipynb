```python
import joblib
from pathlib import Path
from fastapi import FastAPI
from fastapi.testclient import TestClient
from pydantic import BaseModel,Field
import numpy as np



class Student(BaseModel):
    math1:float=Field(ge=0,re=0)
    math2:float=Field(ge=0,re=0)
    math3:float=Field(ge=0,re=0)
    math4:float=Field(ge=0,re=0)


model=joblib.load("models/model1.joblib")
app=FastAPI()
@app.get("/")
def home():
    return {
        "message":"hiiiiiii"
    }
@app.post("/predict")
def predict(student:Student):
    student_scores=[[
        student.math1,
        student.math2,
        student.math3,
        student.math4
    ]]
    prediction=model.predict(student_scores)

    return {
        "scores": student_scores,
        "pred_final":float( prediction[0])

    }

student1={
    "math1":13,
    "math2":14,
    "math3":20,
    "math4":20
}

client=TestClient(app)

response=client.post("/predict",json=student1)
print(response.json())
```
